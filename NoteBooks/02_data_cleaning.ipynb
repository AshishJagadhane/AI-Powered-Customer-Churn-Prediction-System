```python
import numpy as np
import pandas as pd

df = pd.read_csv("Data/Raw/customer_churn_data.csv")
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
missing_percentage
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
df.drop(columns=["satifaction_score"], inplace=True)
print(df.shape)
```

```python
df.isnull().sum()
```

```python
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())
```

```python
df["monthly_charges"] = df["monthly_charges"].fillna(
    df["monthly_charges"].median()
)

df["total_charges"] = df["total_charges"].fillna(
    df["total_charges"].median()
)
```

```python
df["payment_method"] = df["payment_method"].fillna("Unknown")

df["internet_service"] = df["internet_service"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())
```

```python
import os

os.makedirs("Data/Processed", exist_ok=True)

df.to_csv(
    "Data/Processed/customer_churn_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
